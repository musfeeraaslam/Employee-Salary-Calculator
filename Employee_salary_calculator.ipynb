```python
def calculate_salary(basic_salary, overtime_hours):
  overtime_pay=  overtime_hours*500
  if overtime_hours > 20:
    bonus = 5000
  else:
    bonus = 0

  final_salary = basic_salary + overtime_pay + bonus

  print(f"Basic Salary: {basic_salary:.0f}")
  print(f"Overtime hours: {overtime_hours:.0f}")
  print(f"Overtime pay: {overtime_pay:.0f}")
  print(f"Bonus: {bonus:.0f}")
  print(f"Final_salary: {final_salary:.0f}")

basic_salary   = float(input("Enter basic salary: "))
overtime_hours = float(input("Enter overtime hours: "))

calculate_salary(basic_salary, overtime_hours)
```
